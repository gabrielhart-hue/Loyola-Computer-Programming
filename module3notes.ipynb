#### we can use an if statement to alter the flow of code. After the word if we include an expression that esolves to true or false. This expression can be an built using comparison and logical operators. If we want one of two things to happen, after our if, we can have an else statement. IF there are multiple routes, we can include some elifs in between the if and else. 

```python
age = input("enter your age")
age_as_num = int(age)

if age_as_num >= 65: 
    print("old enough to retire")
elif age_as_num >= 35:
    print("old enough to be president")
else: 
    print("you are too young")
```

```python
number = int(input("enter a integer"))
if number >= 10:
    if number <= 20:
        print("You are within the range")
    else:
        print("You are above the range")
else:
    print("You are below the range")
```

```python
clothing_stores = ["dior", "burlington", "macy's", "target", "walmart" ]
fav_store = input("What is your favorite clothing store?")
has_store = False
for store in clothing_stores:
    if store.lower() == fav_store.lower() :
        has_store = True
if has_store == True:
    print("I like that store too")
else:
    print("I don't like that store")
```
